## Part 1 · Escape routing

*Predict* the layers needed with 80 µm pads and 12 µm lines and spaces.

```python
print("layers:", p.layers_needed(15, 150, 80, 12, 12))
ws = np.arange(5, 26)
plt.figure()
plt.step(ws, [p.layers_needed(15, 150, 80, w, w) for w in ws], where="mid")
plt.xlabel("line = space (um)"); plt.ylabel("layers to escape 15 rows"); plt.grid(True)
plt.show()
```

## Part 2 · Strain map

*Predict* the fraction of joints within 10% of the corner strain on a 25 mm die.

```python
g = p.bump_grid(25, 0.15)
r = np.hypot(g[:, 0], g[:, 1])
s = p.shear_strain(13.4, 100, r, 0.07)
print(f"{len(g)} joints; corner strain {s.max():.3f}; within 10%: {100*np.mean(r >= 0.9*r.max()):.2f} %")
plt.figure(figsize=(5, 4.2))
plt.scatter(g[::7, 0], g[::7, 1], c=s[::7], s=1, cmap="viridis")
plt.colorbar(label="shear strain"); plt.gca().set_aspect("equal")
plt.xlabel("x (mm)"); plt.ylabel("y (mm)")
plt.show()
```

## Part 3 · Weakest link

*Predict* the package's characteristic life relative to its most-strained joint.

```python
g = p.bump_grid(20, 0.15)
s = p.shear_strain(13.4, 100, np.hypot(g[:, 0], g[:, 1]), 0.07)
sc = p.package_scale(s, 4)
sim = p.simulate_package_life(s, 4, trials=2000)
print(f"formula scale {sc:.4f} (equivalent to {sc**-4:.0f} corner joints)")
print(f"mean first failure: simulated {sim.mean():.4f}, formula {sc*math.gamma(1.25):.4f}")
plt.figure()
plt.hist(sim, bins=50)
plt.xlabel("package life / corner-joint Weibull scale"); plt.ylabel("count"); plt.grid(True)
plt.show()
```

## Part 4 · Power bumps

*Predict* the power bumps for 1,500 W at 0.70 V and 0.2 A per bump.

```python
print(f"power bumps: {p.power_bumps(1500, 0.70, 0.2):.0f}")
v = np.linspace(0.5, 1.0, 51)
plt.figure()
plt.plot(v, [2 * p.power_bumps(1500, x, 0.2) for x in v])
plt.xlabel("supply (V)"); plt.ylabel("power + ground bumps"); plt.grid(True)
plt.show()
```
